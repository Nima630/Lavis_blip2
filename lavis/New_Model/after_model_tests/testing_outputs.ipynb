```python
# import os
# import numpy as np
# import matplotlib.pyplot as plt

# # --------------------------------------
# # 1. Set ROOT experiment folder
# # --------------------------------------
# root = "/home/draiman/Desktop/Nima/Lavis_blip2/lavis/output/BLIP2_test/CAM_FRONT_Qform_mini_Adn_1_debug_1"
# sample_id = "20251202112"
# sample_id_Att = "20251202114"

# benign_dir = os.path.join(root, "benign", sample_id, "result")
# attack_dir = os.path.join(root, "attack", sample_id_Att, "result")

# # --------------------------------------
# # 2. Helper to load benign/attack pairs
# # --------------------------------------

# def load_pair(name):
#     benign_path = os.path.join(benign_dir, f"{name}_e0_i0.npy")
#     attack_path = os.path.join(attack_dir, f"{name}_e0_i0.npy")
#     return np.load(benign_path), np.load(attack_path)

# def plot_histogram(data, title="", xlabel="Value", ylabel="Frequency",
#                    bins=60, color="#1f77b4"):
#     """
#     Plots a clean, professional histogram with axis ticks, grid, labels, and title.
#     """
#     plt.hist(data.flatten(), bins=bins, color=color, alpha=0.85, edgecolor='black')
#     plt.title(title, fontsize=12, fontweight='bold')
#     plt.xlabel(xlabel)
#     plt.ylabel(ylabel)
#     plt.grid(True, linestyle='--', alpha=0.5)
#     plt.tight_layout()

# def plot_histogram_ax(ax, data, title="", bins=60):
#     ax.hist(data.flatten(), bins=bins, color="#1f77b4",
#             alpha=0.85, edgecolor='black')
#     ax.set_title(title, fontsize=12, fontweight='bold')
#     ax.set_xlabel("Value")
#     ax.set_ylabel("Frequency")
#     ax.grid(True, linestyle='--', alpha=0.5)


# # Inputs
# cam_input_benign, cam_input_attack = load_pair("cam_input")
# lid_input_benign, lid_input_attack = load_pair("lidar_input")

# # Discrepancy maps
# cam_disc_benign, cam_disc_attack = load_pair("camera_discrepancy")
# lid_disc_benign, lid_disc_attack = load_pair("lidar_discrepancy")

# # --------------------------------------
# # 3. Compute differences
# # --------------------------------------
# cam_input_b_mean = cam_input_benign.mean(0)
# cam_input_a_mean = cam_input_attack.mean(0)
# cam_input_diff   = cam_input_a_mean - cam_input_b_mean

# lid_input_b_mean = lid_input_benign.mean(0)
# lid_input_a_mean = lid_input_attack.mean(0)
# lid_input_diff   = lid_input_a_mean - lid_input_b_mean

# cam_disc_diff = cam_disc_attack - cam_disc_benign
# lid_disc_diff = lid_disc_attack - lid_disc_benign

# # --------------------------------------
# # 4. Create 4-row panel
# # --------------------------------------
# fig, axs = plt.subplots(4, 4, figsize=(18, 16))

# # -------------------------------
# # Row 1 — Camera INPUT
# # -------------------------------
# axs[0,0].imshow(cam_input_b_mean, cmap='viridis'); axs[0,0].set_title("Camera INPUT (Benign)")
# axs[0,1].imshow(cam_input_a_mean, cmap='viridis'); axs[0,1].set_title("Camera INPUT (Attack)")
# axs[0,2].imshow(cam_input_diff, cmap='bwr');       axs[0,2].set_title("Camera INPUT Diff (A-B)")
# # axs[0,3].hist(cam_input_diff.flatten(), bins=60);  axs[0,3].set_title("Camera INPUT Diff Hist.")
# # plt.sca(axs[0,3]) 
# # plot_histogram(
# #     cam_input_diff,
# #     title="Camera INPUT Diff Histogram",
# #     xlabel="Difference",
# #     ylabel="Frequency"
# # )
# plot_histogram_ax(axs[0,3], cam_input_diff, "Camera INPUT Diff Histogram")


# # -------------------------------
# # Row 2 — LiDAR INPUT
# # -------------------------------
# axs[1,0].imshow(lid_input_b_mean, cmap='viridis'); axs[1,0].set_title("LiDAR INPUT (Benign)")
# axs[1,1].imshow(lid_input_a_mean, cmap='viridis'); axs[1,1].set_title("LiDAR INPUT (Attack)")
# axs[1,2].imshow(lid_input_diff, cmap='bwr');       axs[1,2].set_title("LiDAR INPUT Diff (A-B)")
# # axs[1,3].hist(lid_input_diff.flatten(), bins=60);  axs[1,3].set_title("LiDAR INPUT Diff Hist.")
# plt.sca(axs[1,3]) 
# # plot_histogram(
# #     lid_input_diff,
# #     title="LiDAR INPUT Difference Histogram (Attack - Benign)"
# # )
# plot_histogram_ax(axs[1,3], lid_input_diff, "LiDAR INPUT Diff Histogram")



# # -------------------------------
# # Row 3 — Camera DISCREPANCY
# # -------------------------------
# axs[2,0].imshow(cam_disc_benign, cmap='hot');      axs[2,0].set_title("Camera DISC (Benign)")
# axs[2,1].imshow(cam_disc_attack, cmap='hot');      axs[2,1].set_title("Camera DISC (Attack)")
# axs[2,2].imshow(cam_disc_diff, cmap='bwr');        axs[2,2].set_title("Camera DISC Diff (A-B)")
# # axs[2,3].hist(cam_disc_diff.flatten(), bins=60);   axs[2,3].set_title("Camera DISC Diff Hist.")
# plt.sca(axs[2,3]) 
# # plot_histogram(
# #     cam_disc_diff,
# #     title="Camera Discrepancy Difference Histogram (Attack - Benign)"
# # )
# plot_histogram_ax(axs[2,3], cam_disc_diff,  "Camera DISC Diff Histogram")




# # -------------------------------
# # Row 4 — LiDAR DISCREPANCY
# # -------------------------------
# axs[3,0].imshow(lid_disc_benign, cmap='hot');      axs[3,0].set_title("LiDAR DISC (Benign)")
# axs[3,1].imshow(lid_disc_attack, cmap='hot');      axs[3,1].set_title("LiDAR DISC (Attack)")
# axs[3,2].imshow(lid_disc_diff, cmap='bwr');        axs[3,2].set_title("LiDAR DISC Diff (A-B)")
# # axs[3,3].hist(lid_disc_diff.flatten(), bins=60);   axs[3,3].set_title("LiDAR DISC Diff Hist.")
# # plt.sca(axs[3,3]) 
# # plot_histogram(
# #     lid_disc_diff,
# #     title="LiDAR Discrepancy Difference Histogram (Attack - Benign)"
# # )
# plot_histogram_ax(axs[3,3], lid_disc_diff,  "LiDAR DISC Diff Histogram")


# # Remove axes
# for ax in axs.flatten():
#     ax.axis("off")

# plt.tight_layout()
# plt.show()



import os
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------
# 1. Set ROOT experiment folder
# --------------------------------------
root = "/home/draiman/Desktop/Nima/Lavis_blip2/lavis/output/BLIP2_test/CAM_FRONT_Qform_mini_Adn_1_debug_1"
sample_id = "20251202112"
sample_id_Att = "20251202114"

benign_dir = os.path.join(root, "benign", sample_id, "result")
attack_dir = os.path.join(root, "attack", sample_id_Att, "result")

# --------------------------------------
# 2. Helper to load benign/attack pairs
# --------------------------------------

def load_pair(name):
    benign_path = os.path.join(benign_dir, f"{name}_e0_i0.npy")
    attack_path = os.path.join(attack_dir, f"{name}_e0_i0.npy")
    return np.load(benign_path), np.load(attack_path)

def plot_histogram_ax(ax, data, title="", bins=60):
    ax.hist(data.flatten(), bins=bins, color="#1f77b4",
            alpha=0.85, edgecolor='black')
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_xlabel("Value")
    ax.set_ylabel("Frequency")
    ax.grid(True, linestyle='--', alpha=0.5)


# --------------------------------------
# 3. Load inputs + discrepancy maps
# --------------------------------------
cam_input_benign, cam_input_attack = load_pair("cam_input")
lid_input_benign, lid_input_attack = load_pair("lidar_input")

cam_disc_benign, cam_disc_attack = load_pair("camera_discrepancy")
lid_disc_benign, lid_disc_attack = load_pair("lidar_discrepancy")

# --------------------------------------
# 4. Compute differences (CORRECT WAY)
# --------------------------------------
# Subtract first → then mean → accurate
cam_input_diff = cam_input_attack - cam_input_benign
lid_input_diff = lid_input_attack - lid_input_benign

# 2D views (channel-mean)
cam_input_b_mean = cam_input_benign.mean(0)
cam_input_a_mean = cam_input_attack.mean(0)
cam_input_diff2d = cam_input_diff.mean(0)

lid_input_b_mean = lid_input_benign.mean(0)
lid_input_a_mean = lid_input_attack.mean(0)
lid_input_diff2d = lid_input_diff.mean(0)

# Discrepancy difference
cam_disc_diff = cam_disc_attack - cam_disc_benign
lid_disc_diff = lid_disc_attack - lid_disc_benign

# --------------------------------------
# 5. Create 4-row panel
# --------------------------------------
fig, axs = plt.subplots(4, 4, figsize=(18, 16))

# -------------------------------
# Row 1 — Camera INPUT
# -------------------------------
axs[0,0].imshow(cam_input_b_mean, cmap='viridis'); axs[0,0].set_title("Camera INPUT (Benign)")
axs[0,1].imshow(cam_input_a_mean, cmap='viridis'); axs[0,1].set_title("Camera INPUT (Attack)")
axs[0,2].imshow(cam_input_diff2d, cmap='bwr');     axs[0,2].set_title("Camera INPUT Diff (A-B)")
plot_histogram_ax(axs[0,3], cam_input_diff, "Camera INPUT Diff Histogram")

# -------------------------------
# Row 2 — LiDAR INPUT
# -------------------------------
axs[1,0].imshow(lid_input_b_mean, cmap='viridis'); axs[1,0].set_title("LiDAR INPUT (Benign)")
axs[1,1].imshow(lid_input_a_mean, cmap='viridis'); axs[1,1].set_title("LiDAR INPUT (Attack)")
axs[1,2].imshow(lid_input_diff2d, cmap='bwr');     axs[1,2].set_title("LiDAR INPUT Diff (A-B)")
plot_histogram_ax(axs[1,3], lid_input_diff, "LiDAR INPUT Diff Histogram")

# -------------------------------
# Row 3 — Camera DISCREPANCY
# -------------------------------
axs[2,0].imshow(cam_disc_benign, cmap='hot');      axs[2,0].set_title("Camera DISC (Benign)")
axs[2,1].imshow(cam_disc_attack, cmap='hot');      axs[2,1].set_title("Camera DISC (Attack)")
axs[2,2].imshow(cam_disc_diff, cmap='bwr');        axs[2,2].set_title("Camera DISC Diff (A-B)")
plot_histogram_ax(axs[2,3], cam_disc_diff, "Camera DISC Diff Histogram")

# -------------------------------
# Row 4 — LiDAR DISCREPANCY
# -------------------------------
axs[3,0].imshow(lid_disc_benign, cmap='hot');      axs[3,0].set_title("LiDAR DISC (Benign)")
axs[3,1].imshow(lid_disc_attack, cmap='hot');      axs[3,1].set_title("LiDAR DISC (Attack)")
axs[3,2].imshow(lid_disc_diff, cmap='bwr');        axs[3,2].set_title("LiDAR DISC Diff (A-B)")
plot_histogram_ax(axs[3,3], lid_disc_diff, "LiDAR DISC Diff Histogram")

# Remove axes
for ax in axs.flatten():
    ax.axis("off")

plt.tight_layout()
plt.show()

print("Max diff:", np.abs(cam_input_benign - cam_input_attack).max())
print("Sum diff:", np.abs(cam_input_benign - cam_input_attack).sum())

```

```python
import os
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------
# 1. Set ROOT experiment folder
# --------------------------------------
root = "/home/draiman/Desktop/Nima/Lavis_blip2/lavis/output/BLIP2_test/CAM_FRONT_Qform_mini_Adn_1_debug_1"
sample_id_benign = "20251202135"
sample_id_attack = "20251202135"   

benign_dir = os.path.join(root, "benign", sample_id_benign, "result")
attack_dir = os.path.join(root, "attack", sample_id_attack, "result")

# --------------------------------------
# 2. Helper to load benign/attack pairs
# --------------------------------------
def load_pair(name):
    benign_path = os.path.join(benign_dir, f"{name}_e0_i0.npy")
    attack_path = os.path.join(attack_dir, f"{name}_e0_i0.npy")
    return np.load(benign_path), np.load(attack_path)

def plot_histogram_ax(ax, data, title="", bins=60):
    ax.hist(data.flatten(), bins=bins, color="#1f77b4",
            alpha=0.85, edgecolor='black')
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_xlabel("Value")
    ax.set_ylabel("Frequency")
    ax.grid(True, linestyle='--', alpha=0.5)


# --------------------------------------
# 3. Load INPUTS + RECONSTRUCTIONS
# --------------------------------------
cam_input_b, cam_input_a = load_pair("cam_input")
lid_input_b, lid_input_a = load_pair("lidar_input")

cam_rec_b, cam_rec_a     = load_pair("cam_rec")
lid_rec_b, lid_rec_a     = load_pair("lid_rec")

# --------------------------------------
# 4. Compute Recon - Input
# --------------------------------------
cam_err_b = cam_rec_b - cam_input_b      # benign reconstruction error
cam_err_a = cam_rec_a - cam_input_a      # attack reconstruction error
cam_err_diff = cam_err_a - cam_err_b     # difference in errors

lid_err_b = lid_rec_b - lid_input_b
lid_err_a = lid_rec_a - lid_input_a
lid_err_diff = lid_err_a - lid_err_b

# 2D channel-mean for visualization
cam_err_b_2d = cam_err_b.mean(0)
cam_err_a_2d = cam_err_a.mean(0)
cam_err_diff2d = cam_err_diff.mean(0)

lid_err_b_2d = lid_err_b.mean(0)
lid_err_a_2d = lid_err_a.mean(0)
lid_err_diff2d = lid_err_diff.mean(0)

# --------------------------------------
# 5. Create 4-row visualization
# --------------------------------------
fig, axs = plt.subplots(4, 4, figsize=(20, 18))

# -------------------------------
# Row 1 — Camera INPUT
# -------------------------------
axs[0,0].imshow(cam_input_b.mean(0), cmap='viridis'); axs[0,0].set_title("Camera INPUT (Benign)")
axs[0,1].imshow(cam_input_a.mean(0), cmap='viridis'); axs[0,1].set_title("Camera INPUT (Attack)")
axs[0,2].imshow((cam_input_a - cam_input_b).mean(0), cmap='bwr'); axs[0,2].set_title("Camera INPUT Diff (A-B)")
plot_histogram_ax(axs[0,3], cam_input_a - cam_input_b, "Camera INPUT Diff Hist.")

# -------------------------------
# Row 2 — Camera Reconstruction Error
# -------------------------------
axs[1,0].imshow(cam_err_b_2d, cmap='hot'); axs[1,0].set_title("Camera RECON-INPUT (Benign)")
axs[1,1].imshow(cam_err_a_2d, cmap='hot'); axs[1,1].set_title("Camera RECON-INPUT (Attack)")
axs[1,2].imshow(cam_err_diff2d, cmap='bwr'); axs[1,2].set_title("Camera Error Diff (Attack - Benign)")
plot_histogram_ax(axs[1,3], cam_err_diff, "Camera RECON Error Diff Hist.")

# -------------------------------
# Row 3 — LiDAR INPUT
# -------------------------------
axs[2,0].imshow(lid_input_b.mean(0), cmap='viridis'); axs[2,0].set_title("LiDAR INPUT (Benign)")
axs[2,1].imshow(lid_input_a.mean(0), cmap='viridis'); axs[2,1].set_title("LiDAR INPUT (Attack)")
axs[2,2].imshow((lid_input_a - lid_input_b).mean(0), cmap='bwr'); axs[2,2].set_title("LiDAR INPUT Diff (A-B)")
plot_histogram_ax(axs[2,3], lid_input_a - lid_input_b, "LiDAR INPUT Diff Hist.")

# -------------------------------
# Row 4 — LiDAR Reconstruction Error
# -------------------------------
axs[3,0].imshow(lid_err_b_2d, cmap='hot'); axs[3,0].set_title("LiDAR RECON-INPUT (Benign)")
axs[3,1].imshow(lid_err_a_2d, cmap='hot'); axs[3,1].set_title("LiDAR RECON-INPUT (Attack)")
axs[3,2].imshow(lid_err_diff2d, cmap='bwr'); axs[3,2].set_title("LiDAR Error Diff (Attack - Benign)")
plot_histogram_ax(axs[3,3], lid_err_diff, "LiDAR RECON Error Diff Hist.")

# Remove axes
for ax in axs.flatten():
    ax.axis("off")

plt.tight_layout()
plt.show()

print("Camera Reconstruction Error (Benign) Sum:", np.abs(cam_err_b).sum())
print("Camera Reconstruction Error (Attack) Sum:", np.abs(cam_err_a).sum())
print("LiDAR Reconstruction Error (Benign) Sum:", np.abs(lid_err_b).sum())
print("LiDAR Reconstruction Error (Attack) Sum:", np.abs(lid_err_a).sum())
```
