```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# ===== Load file =====
abs_path = "/home/draiman/Desktop/debug_cam_features/cam_features.pt"
feat = torch.load(abs_path, map_location="cpu")  # shape [256, H, W]

feat = feat.numpy()    # convert to numpy
channel = feat[0]      # select channel 0 → shape [H, W]

plt.figure(figsize=(8, 8))
plt.imshow(channel, cmap="inferno")
plt.colorbar()
plt.title("Camera BEV Feature Heatmap (Channel 0)")
plt.show()
```
